```python

fila = []

def adicionar(nome, idade, prioridade):
    fila.append([nome, idade, prioridade])

def atender():
    if fila:
        paciente = fila.pop(0)
        print("Atendendo:", paciente)
    else:
        print("Fila vazia")

def listar():
    print("Fila de espera:")
    for paciente in fila:
        print(paciente)

def vazia():
    return len(fila) == 0


adicionar("João", 20, "Normal")
adicionar("Maria", 35, "Urgente")
adicionar("Pedro", 50, "Normal")

listar()

atender()

listar()

print("Está vazia?", vazia())
```
