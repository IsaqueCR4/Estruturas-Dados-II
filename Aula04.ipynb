```python
pilha = []


pilha.append(10)
pilha.append(20)
pilha.append(30)
pilha.append(40)

soma = 0

while pilha:
    numero = pilha.pop()
    print("Removendo:", numero)
    soma += numero

print("Soma total:", soma)
```

```python
fila = []

while True:
    nome = input("Digite o nome do cliente: ")

    if nome == "sair":
        break

    fila.append(nome)

    print("Fila:", fila)

    cliente = fila.pop(0)

    print("Atendendo:", cliente)
    print()
```

01/ Pilha via Array Estático.

```python
#include <stdio.h>

#define TAMANHO 5

int pilha[TAMANHO];
int top = -1;

// Adicionar elemento
void push(int valor) {
    if (top == TAMANHO - 1) {
        printf("Pilha cheia!\n");
    } else {
        top++;
        pilha[top] = valor;
    }
}

int pop() {
    if (top == -1) {
        printf("Pilha vazia!\n");
        return -1;
    } else {
        int valor = pilha[top];
        top--;
        return valor;
    }
}

int main() {

    push(10);
    push(20);
    push(30);

    printf("Removido: %d\n", pop());
    printf("Removido: %d\n", pop());
    printf("Removido: %d\n", pop());

    return 0;
}
```

02/Fila Circular

```python
#include <stdio.h>

#define TAMANHO 5

int fila[TAMANHO];
int inicio = 0;
int fim = 0;

void enqueue(int valor) {
    if (fim == TAMANHO) {
        printf("Fila cheia!\n");
    } else {
        fila[fim] = valor;
        fim++;
    }
}

void dequeue() {
    if (inicio == fim) {
        printf("Fila vazia!\n");
    } else {
        printf("Removido: %d\n", fila[inicio]);
        inicio++;
    }
}

int main() {
    enqueue(10);
    enqueue(20);
    enqueue(30);

    dequeue();
    dequeue();

    enqueue(40);
    enqueue(50);

    dequeue();
    dequeue();
    dequeue();

    return 0;
}


```
